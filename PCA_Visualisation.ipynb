```python
# -*- coding: utf-8 -*-
"""PCA Analysis on Genetic Data"""

import pandas as pd
from sklearn.decomposition import PCA
import altair as alt

# Load data
df = pd.read_csv('matrix.csv')

# Rename first column if necessary
df.rename(columns={'Unnamed: 0': 'Sample'}, inplace=True)

# Exclude non-SNP columns
non_snp_columns = ['Population code', 'Sample']

# Select only SNP (numeric) data for PCA
df_snps = df.drop(columns=non_snp_columns, axis=1)

# Convert to numpy array
matrix = df_snps.to_numpy(dtype=float)  # Convert data to float type

# Apply PCA
pca = PCA(n_components=2)
to_plot = pca.fit_transform(matrix)

# Load population metadata
pop = pd.read_csv('igsr_populations.tsv', sep='\t')

# Merge PCA results with population data
df_plot = df[non_snp_columns].copy()
df_plot['PC1'] = to_plot[:, 0]
df_plot['PC2'] = to_plot[:, 1]

# Merge with population metadata
df_plot = df_plot.merge(pop, on='Population code', how='inner')

# Altair visualization (Only the correct plot remains)
chart = alt.Chart(df_plot).mark_point().encode(
    x='PC1',
    y='PC2',
    color=alt.Color('Population code', scale=alt.Scale(scheme='category20'))
)

chart

```
